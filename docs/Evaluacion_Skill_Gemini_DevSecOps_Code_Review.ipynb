El conjunto de datos incluido en este cuaderno es **sintético y demostrativo**: sus valores se diseñaron para ilustrar el procedimiento de análisis y no corresponden a resultados medidos directamente en Gemini. Para presentar resultados empíricos en un informe académico, se deben sustituir por registros reales obtenidos mediante un protocolo de prueba documentado. La precisión OWASP representa el porcentaje estimado de hallazgos correctos en este ejemplo; una evaluación real debe definir el conjunto de vulnerabilidades de referencia y cómo se calculan verdaderos positivos, falsos positivos y falsos negativos.

## Estructura del análisis

1. Importación de bibliotecas y configuración visual.
2. Construcción del DataFrame con diez ejecuciones de prueba.
3. Resumen estadístico de las métricas cuantitativas.
4. Comparación visual de precisión y tiempos de respuesta.
5. Conclusiones basadas en los resultados calculados.

```python
# 1. IMPORTACIÓN DE BIBLIOTECAS
# pandas: estructura y análisis tabular de los registros.
# numpy: operaciones numéricas y apoyo para cálculos estadísticos.
# matplotlib.pyplot: creación y personalización de figuras.
# seaborn: visualizaciones estadísticas con una estética consistente.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Establece un tema legible para los gráficos y un tamaño base de figura.
sns.set_theme(style='darkgrid')
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['font.size'] = 10

print('Bibliotecas importadas y configuración visual aplicada correctamente.')
```

```python
# 2. CREACIÓN DEL DATASET DE PRUEBA
# Cada fila representa una ejecución de evaluación de la Skill.
# Los datos son sintéticos y se usan para demostrar el flujo de análisis.
# Los tiempos están expresados en segundos y la precisión en porcentaje.

datos_pruebas = {
    'ID_Prueba': list(range(1, 11)),
    'Lenguaje': [
        'Python', 'Python', 'C#', 'C#', 'JavaScript',
        'JavaScript', 'TypeScript', 'TypeScript', 'SQL', 'SQL'
    ],
    'Modulo_Evaluado': [
        'Auth API', 'AgroConnect Backend', 'Vehicle Wash Logic', 'Auth API',
        'AgroConnect Backend', 'Vehicle Wash Logic', 'Auth API', 'AgroConnect Backend',
        'Módulo de consultas SQL', 'Módulo de reportes SQL'
    ],
    'Tiempo_Respuesta_Seg': [1.7, 2.1, 2.4, 2.0, 1.8, 2.6, 1.9, 2.2, 1.4, 1.9],
    'Precision_OWASP_Pct': [96, 94, 91, 93, 97, 90, 98, 95, 100, 94],
    'Refactorizacion_SOLID': [True] * 10,
    'Pruebas_Unitarias_Generadas': [8, 10, 7, 9, 11, 8, 12, 10, 6, 9]
}

# Convierte el diccionario en un DataFrame, estructura tabular de pandas.
df = pd.DataFrame(datos_pruebas)

# Comprobaciones básicas para prevenir errores en el análisis posterior.
assert len(df) == 10, 'El dataset debe contener exactamente 10 ejecuciones.'
assert df['Tiempo_Respuesta_Seg'].between(1.4, 2.6).all(), 'Hay tiempos fuera del rango esperado.'
assert df['Precision_OWASP_Pct'].between(88, 100).all(), 'Hay porcentajes fuera del rango esperado.'
assert df['Refactorizacion_SOLID'].dtype == bool, 'La columna SOLID debe ser booleana.'

print('Vista general del dataset:')
display(df)

print('\nTipos de datos y valores no nulos:')
df.info()
```

```python
# 3. ANÁLISIS ESTADÍSTICO DESCRIPTIVO
# Seleccionamos las métricas cuantitativas que admiten resumen numérico.
# ID_Prueba se excluye porque es un identificador, no una métrica de rendimiento.
# Precision_OWASP_Pct, tiempo y número de pruebas generadas sí se resumen.

columnas_metricas = [
    'Tiempo_Respuesta_Seg',
    'Precision_OWASP_Pct',
    'Pruebas_Unitarias_Generadas'
]

resumen = df[columnas_metricas].describe().T
resumen = resumen.rename(columns={
    'count': 'Cantidad',
    'mean': 'Promedio',
    'std': 'Desviación estándar',
    'min': 'Mínimo',
    '25%': 'Percentil 25',
    '50%': 'Mediana',
    '75%': 'Percentil 75',
    'max': 'Máximo'
})

# Redondeamos solo para facilitar la lectura; los cálculos internos conservan precisión.
print('Resumen estadístico de las métricas cuantitativas:')
display(resumen.round(2))

# Indicadores globales complementarios, útiles para redactar las conclusiones.
tiempo_promedio = df['Tiempo_Respuesta_Seg'].mean()
precision_promedio = df['Precision_OWASP_Pct'].mean()
pruebas_promedio = df['Pruebas_Unitarias_Generadas'].mean()
porcentaje_solid = df['Refactorizacion_SOLID'].mean() * 100

print(f'Tiempo de respuesta promedio: {tiempo_promedio:.2f} segundos')
print(f'Precisión OWASP promedio: {precision_promedio:.2f}%')
print(f'Pruebas unitarias generadas por ejecución (promedio): {pruebas_promedio:.2f}')
print(f'Ejecuciones con refactorización SOLID marcada como exitosa: {porcentaje_solid:.1f}%')
```

```python
# 4. VISUALIZACIONES COMPARATIVAS
# Se construye una figura con dos paneles en una sola fila.
# Panel izquierdo: precisión OWASP promedio por lenguaje.
# Panel derecho: distribución de tiempos de respuesta por lenguaje.

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

# GRÁFICO 1: barras de precisión promedio por lenguaje.
precision_por_lenguaje = (
    df.groupby('Lenguaje', as_index=False)['Precision_OWASP_Pct']
      .mean()
      .sort_values('Precision_OWASP_Pct', ascending=False)
)

sns.barplot(
    data=precision_por_lenguaje,
    x='Lenguaje',
    y='Precision_OWASP_Pct',
    ax=axes[0]
)
axes[0].set_title('Precisión OWASP promedio por lenguaje')
axes[0].set_xlabel('Lenguaje de programación')
axes[0].set_ylabel('Precisión promedio (%)')
axes[0].set_ylim(0, 110)
axes[0].tick_params(axis='x', rotation=20)

# Coloca la etiqueta porcentual ligeramente por encima de cada barra.
for barra in axes[0].patches:
    altura = barra.get_height()
    axes[0].annotate(
        f'{altura:.1f}%',
        (barra.get_x() + barra.get_width() / 2, altura),
        ha='center', va='bottom', fontsize=9,
        xytext=(0, 4), textcoords='offset points'
    )

# GRÁFICO 2: diagrama de caja para comparar tiempos y dispersión.
orden_lenguajes = ['Python', 'C#', 'JavaScript', 'TypeScript', 'SQL']
sns.boxplot(
    data=df,
    x='Lenguaje',
    y='Tiempo_Respuesta_Seg',
    order=orden_lenguajes,
    ax=axes[1]
)
axes[1].set_title('Distribución del tiempo de respuesta por lenguaje')
axes[1].set_xlabel('Lenguaje de programación')
axes[1].set_ylabel('Tiempo de respuesta (segundos)')
axes[1].tick_params(axis='x', rotation=20)

fig.suptitle('Evaluación de la Skill personalizada para DevSecOps y Code Review', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

# Tabla auxiliar con promedios por lenguaje para facilitar la interpretación.
comparacion = df.groupby('Lenguaje').agg(
    Tiempo_Promedio_Seg=('Tiempo_Respuesta_Seg', 'mean'),
    Precision_Promedio_Pct=('Precision_OWASP_Pct', 'mean'),
    Pruebas_Unitarias_Promedio=('Pruebas_Unitarias_Generadas', 'mean')
).reindex(orden_lenguajes)

print('Comparación numérica por lenguaje:')
display(comparacion.round(2))
```

# 5. Conclusiones técnicas

Las siguientes conclusiones se derivan de los diez registros sintéticos definidos en este cuaderno. Deben interpretarse como resultados demostrativos, no como una validación real de Gemini.

1. **Tiempo de respuesta:** el tiempo promedio de las diez ejecuciones es de aproximadamente **2,00 segundos**, dentro del intervalo definido de 1,4 a 2,6 segundos. Para una evaluación real conviene ejecutar cada escenario varias veces y reportar también percentiles (por ejemplo, P50 y P95), ya que el promedio por sí solo no refleja toda la variabilidad.

2. **Precisión en detección de riesgos OWASP:** la precisión promedio del dataset es de aproximadamente **94,8 %**, lo que representa un desempeño alto dentro de este ejemplo. Sin embargo, este porcentaje solo será defendible como métrica de seguridad si se calcula comparando los hallazgos de la Skill con un conjunto de referencia revisado por expertos y se documentan falsos positivos y falsos negativos.

3. **Apoyo a calidad y pruebas:** las diez ejecuciones tienen `Refactorizacion_SOLID = True` en el dataset, equivalente al **100 % de los registros marcados como satisfactorios**, y se generan en promedio **9 pruebas unitarias por ejecución**. Esta marca es una etiqueta de ejemplo y no demuestra por sí misma que el código refactorizado cumpla SOLID; para comprobarlo se necesita revisión técnica, pruebas automatizadas y criterios de aceptación explícitos.

## Recomendaciones para una evaluación real

- Definir casos de prueba reproducibles para cada lenguaje y módulo, con versiones de herramientas y fecha de ejecución.
- Preparar un conjunto de código con vulnerabilidades conocidas y una solución de referencia validada por una persona experta.
- Medir precisión, recall, F1 y tasa de falsos positivos, además del tiempo de respuesta.
- Repetir las pruebas bajo condiciones comparables y conservar los resultados originales para auditoría.
- Evaluar las recomendaciones SOLID y las pruebas unitarias generadas mediante criterios verificables, no únicamente con una marca booleana.
